### Python Libraries

* A library is a collection of pre-written code that helps you perform tasks faster and more efficiently.
* Instead of writing hundreds of lines of code, you can use a few library functions to get the job done.

1. A library contains ready-made functions and tools that solve common problems. 

Examples:
🔢 Numerical Computing: NumPy   
📊 Data Analysis: Pandas  
📈 Visualization: Matplotlib  
🤖 Machine Learning: Scikit-learn  
🌐 Web Development: Django, Flask

4. Pandas Basics  
* Pandas is the most popular library for data analysis.  
Used for:  
✔️ Reading Excel Files  
✔️ Reading CSV Files  
✔️ Data Cleaning  
✔️ Data Transformation  
✔️ Reporting 
* Install Pandas: pip install pandas 
* Import Pandas: import pandas as pd  

````python
# Create a DataFrame
%pip install pandas
import pandas as pd
data = {
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Price": [50000, 20000, 15000]
}
df = pd.DataFrame(data)
print(df)
````

5. View Data  
* Display first few rows: print(df.head())  
* Display column names: print(df.columns)  
* Display summary: print(df.info())
