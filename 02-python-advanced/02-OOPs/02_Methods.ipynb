#### Methods: 
-----------------------------------------------------------------------------------------------

* **Types of Methods**
* A method is a function defined inside a class that is used to perform a specific operation or define the behavior of an object or class.

* Method = Function inside a class

```
class Employee:

    def display_info(self):
        print("Employee information")

# Employee → Class
# display_info() → Method
# print() → Operation performed by the method

```

* *There are three types of methods in Python:*

* Instance Method — Object-related method --> no decorator
* Class Method    — Class-related method  --> @classmethod
* Static Method   — General utility method --> @staticmethod

1. **Instance Method:**

* If we are using at least one **instance variable** inside a method, it is an instance method, regardless of whether we are also using static/class variables.
* If we are accessing **instance variables** inside a method, it is an instance method, regardless of whether we are also using static/class or local variables.
* The first parameter of an instance method should be `self`.
* self is a reference to the current object.
* Inside a method, if we are trying to access at least one instance variable, then the method is associated with a **particular object**. Hence, it is called an **instance method**.
* The first parameter of an instance method is normally `self`, which is a **reference to the current object**.

````python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    
    def change_grade(self, new_grade):   # instance method
        self.grade = new_grade
        
# Implementation of instance method
student1 = Student("Alice", "A")
new_grade = student1.change_grade("A+")
print(f"{student1.name}'s new grade is: {student1.grade}")  # Output: Alice's new grade is: A+


student2 = Student("Bob", "B")
new_grade = student2.change_grade("C")
print(f"{student2.name}'s new grade is: {student2.grade}")  # Output: Bob's new grade is: C
````

2. **Class Method**

* If we are **not using any instance variables** but are using **class variables**, then we can define the method as a **class method**.
* A class method is defined using the `@classmethod` decorator.
* `cls` is the **reference variable to the current class**.
* The first parameter of a class method is normally `cls`.

### Example

```python
class Employee:

    company = "ABC"

    @classmethod
    def display_company(cls):
        print(cls.company)
```
Here, `display_company()` is a **class method**, and `cls` refers to the `Employee` class.

````python
class School:
    school_name = "ABC School"  # class variable
    
    @classmethod
    def change_school_name(cls, new_name):  # class method
        cls.school_name = new_name

print(School.school_name) 
School.change_school_name("XYZ School") 
print(School.school_name)
````

3. **Static Method**

* If we are not using any instance variables, the method can be a class method or a static method, depending on whether it needs to work with class-level data.
* If we are not using any instance variables or class variables, and the method is not related to a particular object or class, it can be defined as a static method.
* A static method is generally used as a general utility method that performs a specific operation independently of the object and class.
* A static method is defined using the @staticmethod decorator.
* A static method does not require self or cls as its first parameter.

````python
class Calculator: 
    @staticmethod
    def add(a, b):
        return a + b

# A static method can be called using the class name:

result = Calculator.add(10, 20)
print(result)
````

````python
# Staticmethod and Classmethod 
class Student:
    collegeName='JNTU'
    directorname='acmarpu'
    def __init__(self,name,rollno):  # Arguments : self,name,rollno,marks
        self.name = name
        self.rollno = rollno

    def getStudentInfo(self):
        print('Student name:', self.name)
        print('Student roll number:', self.rollno)

    @classmethod
    def getCollegeInfo(cls):
        print('College name:', cls.collegeName)
        print('DIrector name:', cls. directorname)

    @staticmethod
    def getAverage(a,b,c):
        return (a+b+c)/3

s=Student('acmarpu',100)
s.getStudentInfo()
Student.getCollegeInfo()
average=Student.getAverage(10,20,30)
print('Average marks:',average)
````

#### Inheritance
* Inheritance in Python allows a class(subclass) to inherit attributes and methhods from another class(superclass)

* this facilitates code reuse and enables the creation of hierarchical relationships between classes 

````python
# Bank Exmple for single Inheritance

# Account

# Saving Account(Account)

# Parent class: Account

class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def display_account_info(self):
        print(f"Account Number: {self.account_number}")
        print(f"Balance: {self.balance}")


# Child class: Saving Account

class SavingAccount(Account):  # Single Inheritance
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)  # Call the parent class constructor
        self.interest_rate = interest_rate

    def calculate_interest(self):
        return self.balance * self.interest_rate / 100

# Create am instance of SavingAccount
saving_account = SavingAccount("123456789", 1000, 5)
saving_account.display_account_info()
print(f"Interest: {saving_account.calculate_interest()}")  # Output: Interest: 50.0
````
