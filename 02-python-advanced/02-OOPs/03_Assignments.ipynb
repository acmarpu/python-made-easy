```python
# Scenario: A simple Banking System:

# Image we're bulding  a banking system where customers can open accounts, 
# deposit money, withdraw money and view their account balance
# Each bank account  has a unique account number, a balance, and a list of transactions.



#Step 1: Define the Bank Account class: we define a bank account class
# that represents a bank account ,This class will have attrributes for 
# the accont number, balance, and transaction, and methods to perform operations
# like depositing and withdrawing money.

class BankAccount: # Class object 
    def __init__(self, account_number, initial_balance=0):
        self.account_number = account_number
        self.balance = initial_balance
        self.transactions = []

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.transactions.append(f'Deposited: {amount}')
            # print(f'Deposited {amount}. New balance: {self.balance}')
        else:
            print('Deposit amount must be positive.')


    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            self.transactions.append(f'Withdrew: {amount}')
            # print(f'Withdrew {amount}. New balance: {self.balance}')
        else:
            print('Insufficient funds or invalid withdrawal amount.')

    def get_balance(self, account_number):
        if account_number == self.account_number:
            return self.balance
        # else:
        #     print('Account not found.')
        #     return None

    def get_transactions(self, account_number):
        if account_number == self.account_number:
            return self.transactions
        # else:
        #     print('Account not found.')
        #     return None

# Step 2: Create a cuatomer Accounts: Next create instances of the 
# Bank account class for diffrent customers. This demonstrates the 
# instantiation of object from a class 
 
account1 = BankAccount('123456', 1000)  # Instance object
account2 = BankAccount('789012', 5000)  # instance object

# print(dir(account1))

account1.deposit(500)
account1.withdraw(200)
print(f'Account 1 balance: {account1.get_balance("123456")}')
print(f'Account 1 transactions: {account1.get_transactions("123456")}')


#Step 3: Inheritance and Polymorphism: Suppose we later decide to
# Introduce diffrent types of bank accounts, such as saving accounts
# and checking accounts, each with its own rules for deposits and withdrawals.
# we can use inheritance to create child classes from the bank account class,
# and override or extend methods as needed.

class SavingsAccount(BankAccount):
    def __init__(self, account_number, initial_balance=0, interest_rate=0.02):
        super().__init__(account_number, initial_balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        interest = self.balance * self.interest_rate
        self.deposit(interest)  # Reuse the deposit method to add interset

class CheckingAccount(BankAccount):
    def __init__(self, account_number, initial_balance=0, overdraft_limit=500):
        super().__init__(account_number, initial_balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if amount > 0 and (self.balance + self._overdraft_limit) >= amount:
            self.balance -= amount
            self.transactions.append(f'Withdrew: {amount}')
            # print(f'Withdrew {amount}. New balance: {self.balance}')
        else:
            print('Insufficient funds or invalid withdrawal amount.')

# Step 4: Encapsulation and Abstraction : The bank account class encapsulation the 
# data (account number, balance, transactions) and
# the methods (deposit, withdraw, get_balance, get_transactions) that operate
# on the data.
# this encapsulation ensures that the internal state of the bank account is protected,
# and external entities can only interact with it through these methods.
# Abstraction is achived by exposing only the necessary methods
# and holding the implementation details, such as how transactions are 
# stored or how interset is calculated.

# Example usage of the SavingsAccount and CheckingAccount classes 
savings_account = SavingsAccount("111222333", 2000, 0.02)
checking_account = CheckingAccount("444555666", 1500, 300)

savings_account.apply_interest()

print(f"Savings Account Number: {savings_account.account_number}, Balance: {savings_account.get_balance(savings_account.account_number)}")
print(f"Checking Account Number: {checking_account.account_number}, Balance: {checking_account.get_balance(checking_account.account_number)}")
```
