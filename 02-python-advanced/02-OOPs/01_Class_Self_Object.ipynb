### What is OOP?

Python is an object-oriented language, allowing you to structure your code using classes and objects for better organization and reusability.

* Classes and objects are the two core concepts in object-oriented programming.
* A class defines what an object should look like, and an object is created based on that class.
* Organizes code into classes and objects

### Class, Objects and Reference Variable

#### Class

* A class acts as a blueprint, plan, model, or design for creating *objects*(instances).
* A class defines the variables (attributes) and methods (behaviors) that its objects can have.
* A class is a collection of objects. Classes are blueprints for creating objects. A class defines a set of attributes and methods that the created objects (instances) can have.
* In Python, we create a class using the class keyword.
* Class define the properties and behaviors
* Class Objectet (Instance): An object is an instance of a class
* An instance is an individual object created from a class


````python
# Creating a Class

class ClassName:
    # attributes
    # methods
    pass

# class → Keyword used to define a class.
# ClassName → Name of the class.
# : → Indicates the beginning of the class body.
# pass → Used when the class does not contain any code yet.
````

#### Self variable: 
-----------------------------------------------------------------------------------------------

* Self hold the current Object.
* self is a reference variable that always points to the current object.
* Used to acceess attributes and methods of the object on class methods
* Within a Python class, we use self to access the current object.
* The first parameter of the constructor (__init__()) is normally self.
* The first parameter of an instance method is normally self.
* We do not need to provide a value for the self variable explicitly. Python automatically provides the reference to the current object.
* We can use self only within the class to refer to the current object.
* Inside the constructor, we use self to declare and initialize object-related variables (instance variables).
* Inside an instance method, we use self to access the values of instance variables.
* self is not a Python keyword. We can technically use any valid variable name, such as delf or kelf, but it is strongly recommended to use self because it is the standard Python convention.

````python
## Creating Self

class ClassName:
    def __init__(self, attribute1, attribute2):
        self.attribute1 = attribute1
        self.attribute2 = attribute2

# self → Reference to the current object.
# attribute1 and attribute2 → Parameters received by the constructor.
# self.attribute1 and self.attribute2 → Instance variables associated with the current object.
# self.attribute1 = attribute1 → Assigns the value of the attribute1 parameter to the instance variable attribute1.
# self.attribute2 = attribute2 → Assigns the value of the attribute2 parameter to the instance variable attribute2.
````

#### Constructors: __init__()
-----------------------------------------------------------------------------------------------
* A constructor is a special method that is automatically called when an object is created.
* The name of the constructor is always __init__().
* We do not need to call the constructor explicitly. It is automatically executed when we create an object.
* For each object, the constructor is executed once during object creation.
* The main purpose of the constructor is to declare and initialize instance variables.
* __init__ stands for initialization.
* The constructor should have at least one parameter, which is self.
* Within a Python class, the constructor is optional. If we do not define a constructor, Python provides a default constructor automatically.
* We can call __init__() explicitly, but doing so will not create a new object. It will execute the method like a normal method.
* Constructor/method overloading is not directly supported in Python.
* If we define multiple __init__() methods in the same class, Python will consider only the last-defined __init__() method. The previous definitions will be overridden.
   

### Object:
-----------------------------------------------------------------------------------------------
* An object is an instance of a class and represents the actual existence of the class in memory.
* For any class, we can create any number of objects.
* Each object has its own instance variables and state.


````python
# Example 1

class Student:
    def __init__(self):
        self.name = "acmarpu"
        self.rollno = 101
        self.marks = 90

    def talk(self):
        print('Hello i am:', self.name)
        print('Hello roolnumber:', self.rollno)
        print('Hello i am marks:', self.marks)

s = Student()
print('Student Name:', s.name)
print('Student Rool Number:',s.rollno)
print('Student marks:', s.marks)
s.talk()


# class Student: , Student: is the class name
# Student() creation of object
# s is the reference_variable
# def talk(self):  methods
# __init__ constructure 

# Student()
# 1.We are creating an object .PVM will create object and allocate memory for that object.
# 2.PVM will execute __init__()
#    to declare and initialize variables of the object (instance variables)

# s = Student()
#      - For newly created and alredy initilized object, we are assigning S reference variable. By using
#        this S, we can access variables and methods on that object.
````

````python
# Example 2

class Student:
    def __init__(self,name,rollno,marks):  # Arguments : self,name,rollno,marks
        self.name = name
        self.rollno = rollno
        self.marks = marks

    def talk(self):
        print('Hello i am:', self.name)
        print('Hello roolnumber:', self.rollno)
        print('Hello i am marks:', self.marks)

s1 = Student('ashoka',101,90)           # Values : 'ashoka',101,90
print('Student1:',s1.name,s1.rollno,s1.marks)

s2 = Student('marpu',102,95)
print('Student2:',s2.name,s2.rollno,s2.marks)
````

````python
# Example 3

class Bankaccount:               # Class Name
    bank_name = "ABC Bank"       # Class variable
    branch_name = "Main Branch"  # Class variable

    def __init__(self, account_number, account_holder_name, balance):
        self.account_number = account_number            # Instance Variable
        self.account_holder_name = account_holder_name  # Instance Variable
        self.balance = balance                          # Instance Variable  

    def display_account_info(self):
        print("Account Number:", self.account_number)
        print("Account Holder Name:", self.account_holder_name)
        print("Balance:", self.balance)

# Creating instance object
Account1 = Bankaccount("123456789", "marpu", 1000) # Instance object Account1
Account2 = Bankaccount("987654321", "ac", 5000)
Account3 = Bankaccount("456789123", "ashoka", 2000)

# Displaying account information for each account
print("Account 1 Information:", Account1.display_account_info())
print("Account 2 Information:", Account2.display_account_info())
print("Account 3 Information:", Account3.display_account_info())
print("Bank Name:", Bankaccount.bank_name)                     # Accessing class variable
print("Branch Name:", Bankaccount.branch_name)                 # Accessing class variable
````

Types of variables:
     - 3 types
     - 1. Instance Variables/ Object Level variables
     - 2. Static variables/ Class Level Variables
     - 3. Local Variables/ Method level Variables

     - Instance variable+Static, variable+local variable ---> instance method
     - static variables + local variables ---> class method
     - local variables -----> static method
     

     - 1. Instance Variables :
       - Object level variables
       - if the value of a variable is varied form object to object, such type of variables are called instance variables
       - for every object a separate copy will be created.
       - in general we can define intsnace variables inside constructor by using self.

         ```
       class Student:
           collegeName= 'JNTU'    
           def __init__(self,name,rollno)
               self.name=name                # Instance Variables
               self.rollno=rollno            # Instance Variables
      ```


     - 2. Static variables
       - if the value of a variable is not varied from object to object then it is not recommended to declared that variabl as instance variable. static variables/class level variables
       - in thecase of instance variables, for every object a separate copy. But in the case of static variables, a single copy will be created and shared to every object of that class
       - Most of the times, static variables should be declread within the class directly.
       - in general we can declare static variables within the class directly, but outside of any method.

      ```
      class Test:
          a=10
          def __init__(self):
              Test.b=20
          
          def m1(self):
              Test.c=30
          
          @classmethod
          def m2(cls):
              Test.d=40
              cls.e=50
          
          @staticmethod
          def m3():
              Test.f=60

      t=Test()
      t.m1()
      Test.m2()
      Test.m3()
      Test.g=70
      print(Test.__dict__)

      ```

       - local variables will be created at the time of method execution and destroyed once method completes.

