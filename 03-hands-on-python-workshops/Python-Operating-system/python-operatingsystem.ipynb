```python
# Get date and time information of given file
import os
import os.path
import datetime
import time

filename = input("Enter the filename: ")
if os.path.exists(filename):
    if os.path.isfile(filename):
        create=os.path.getctime(filename)
        modified=os.path.getmtime(filename)
        access=os.path.getatime(filename)
        print(time.ctime(create))
        print(time.ctime(modified))
        print(f"File created on: {datetime.datetime.fromtimestamp(create)}")
        print(f"File last modified on: {datetime.datetime.fromtimestamp(modified)}")
        print(f"File last accessed on: {datetime.datetime.fromtimestamp(access)}")
    else:
        print("The specified path is not a file.")
else:
    print("File not found.")

# Enter input format  d:\\AI\FB_IMG_1755758351007.jpg
```

```python
# List all the file created in a given month

import os
import os.path
import datetime
import time

month = int(input("Enter the month (1-12): "))
foldername= input("Enter the folder path: ")

if os.path.exists(foldername):
    if os.path.isdir(foldername):
        A=os.listdir(foldername)
        os.chdir(foldername)
        
        for name in A:
            create=os.path.getctime(name)
            dt=datetime.datetime.fromtimestamp(create)
            if dt.date().month == month:
             print(f'{name}----------->{dt}')

        else:
            print("The specified path is not a directory.")
else:
    print("Directory not found.")
        




```

```python
import os
import os.path
import datetime
import time
import shutil

month = int(input("Enter the month (1-12): "))
foldername = input("Enter the folder path: ")

dict1 = {
    1:'January', 2:'February', 3:'March', 4:'April',
    5:'May', 6:'June', 7:'July', 8:'August',
    9:'September', 10:'October', 11:'November', 12:'December'
}

if os.path.exists(foldername):

    if os.path.isdir(foldername):

        A = os.listdir(foldername)
        os.chdir(foldername)
        os.makedirs(dict1[month], exist_ok=True)

        for name in A:
            create = os.path.getctime(name)
            dt = datetime.datetime.fromtimestamp(create)

            if dt.date().month == month and os.path.isfile(name):
                print(f'{name}----------->{dt.date()}')
                shutil.copy(name, dict1[month])

    else:
        print("The specified path is not a directory.")

else:
    print("Directory not found.")
```
